# Data experimentation of combining the tables to one-big-table
**Motivation:** How can the tables be connected to simplify analysis? What kind of values can be used?  
$ \rightarrow $ 'LoggTogID' seem like a good fit as the combination of 'Tognummer' and date is a unique combination

## 1st idea: combine Terminallogg R25 and Terminallogg R26 into one table for all terminal logs

### Combining the tables

```python
# Confirm that both dataframes have the same columns and column order
if not df_terminallogg_r25.columns.equals(df_terminallogg_r26.columns):
    raise ValueError("The dataframes do not have identical columns.")

# Add the source period before combining
df_terminallogg_r25 = df_terminallogg_r25.assign(periode="R25")
df_terminallogg_r26 = df_terminallogg_r26.assign(periode="R26")

# Combine rows
df_terminallogg = pd.concat(
    [df_terminallogg_r25, df_terminallogg_r26],
    axis=0,
    ignore_index=True
)

# Optional: sort chronologically
df_terminallogg["Avgangsdato"] = pd.to_datetime(
    df_terminallogg["Avgangsdato"],
    errors="coerce"
)

df_terminallogg = df_terminallogg.sort_values(
    "Avgangsdato",
    na_position="last"
).reset_index(drop=True)
```

```python
print(df_terminallogg.shape)
print(df_terminallogg["periode"].value_counts())
print(df_terminallogg["Avgangsdato"].min())
print(df_terminallogg["Avgangsdato"].max())

# Check whether LoggTogID remains unique
duplicate_ids = df_terminallogg[
    df_terminallogg["LoggTogID"].duplicated(keep=False)
]

print(f"Duplicate LoggTogID values: {duplicate_ids['LoggTogID'].nunique()}")
```

## 2nd idea: simplify analysis by creating a one-big-table with the most important information and the grain as "one row per each outgoing departure-train" 

Format of 'LoggTogID': 'DD-MM-YYYY_tognummer'

```python
# Create the LoggTogID for the df_TIOS_delays_report dataset
df_TIOS_delays_report["LoggTogID"] = (
    pd.to_datetime(df_TIOS_delays_report["Dato"]).dt.strftime("%d-%m-%Y")
    + "_"
    + df_TIOS_delays_report["Tognr"].astype(str)
)

df_TIOS_delays_report["LoggTogID"]
```

```python
# Create the LoggTogID for the df_vy_alnabru_timings dataset
df_vy_alnabru_timings["LoggTogID"] = (
    pd.to_datetime(df_vy_alnabru_timings["train_date"]).dt.strftime("%d-%m-%Y")
    + "_"
    + df_vy_alnabru_timings["train_id"].astype(str)
)

df_vy_alnabru_timings["LoggTogID"]
```

```python
# Compare total number of LoggTogIDs with the number of unique LoggTogIDs
dataframes = {
    "df_terminallogg": df_terminallogg,
    "df_TIOS_delays_report": df_TIOS_delays_report,
    "df_vy_alnabru_timings": df_vy_alnabru_timings,
}

loggtogid_summary = pd.DataFrame([
    {
        "DataFrame": name,
        "Number of LoggTogIDs": len(df["LoggTogID"].dropna()),
        "Unique LoggTogIDs": df["LoggTogID"].nunique(),
        "Duplicate LoggTogIDs": (
            len(df["LoggTogID"].dropna()) - df["LoggTogID"].nunique()
        ),
    }
    for name, df in dataframes.items()
])

display(loggtogid_summary)
```

*Comment:* The constructed LoggTogIDs are mostly unique. Duplicated just need to be handled

### Terminallogg $\leftarrow$ TIOS delay report

```python
# Work on copies
terminal = df_terminallogg.copy()
tios = df_TIOS_delays_report.copy()

# Standardize the join key
terminal["LoggTogID"] = (
    terminal["LoggTogID"]
    .astype("string")
    .str.strip()
)

tios["LoggTogID"] = (
    tios["LoggTogID"]
    .astype("string")
    .str.strip()
)

# Remove rows without a join key
terminal = terminal.dropna(subset=["LoggTogID"])
tios = tios.dropna(subset=["LoggTogID"])

# Remove duplicate LoggTogIDs, keeping the first occurrence
terminal = terminal.drop_duplicates(
    subset="LoggTogID",
    keep="first",
)

tios = tios.drop_duplicates(
    subset="LoggTogID",
    keep="first",
)
```

```python
terminal_columns = [
    "LoggTogID",
    "Avgangsdato",
    "Tognummer",
    "Lastefrist",
    "Ferdig opplastet",
    "Frist lok",
    "Lok ankommet",
    "Planlagt klart fra terminal",
    "Klart fra terminal",
    "Planlagt avgang",
    "Faktisk avgang",
    "Årsakskode - avgangsforsinkelse",
    "Kommentar",
]

tios_columns = [
    "LoggTogID",
    "Stasjon",
    "Innstilt",
    "Forsinkelse",
    "Årsak",
    "HL-beskrivelse",
    "Kommentar",
]

terminal = terminal[terminal_columns].rename(columns={
    "Årsakskode - avgangsforsinkelse": "Årsakskode - terminal",
    "Kommentar": "Kommentar - terminal",
})

tios = tios[tios_columns].rename(columns={
    "Årsak": "Årsakskode - BN",
    "HL-beskrivelse": "Kommentar1 - BN",
    "Kommentar": "Kommentar2 - BN",
})

df_terminal_tios = terminal.merge(
    tios,
    on="LoggTogID",
    how="left",
    validate="one_to_one",
)
```

```python
# Display 20 random rows
display(df_terminal_tios.sample(n=20, random_state=37))
```

### Terminallogg $\leftarrow$ TIOS delay report $\leftarrow$ Vy Alnabru timings

```python
# Work on a copy of the Vy data
vy = df_vy_alnabru_timings.copy()

# Standardize the join key in both dataframes
df_terminal_tios["LoggTogID"] = (
    df_terminal_tios["LoggTogID"]
    .astype("string")
    .str.strip()
)

vy["LoggTogID"] = (
    vy["LoggTogID"]
    .astype("string")
    .str.strip()
)

# Count duplicate Vy IDs before removing them
print(
    "Duplicate Vy LoggTogIDs removed:",
    vy["LoggTogID"].duplicated().sum()
)

# Remove rows without a join key
vy = vy.dropna(subset=["LoggTogID"])

# Keep only the first row for each LoggTogID
vy = vy.drop_duplicates(
    subset="LoggTogID",
    keep="first",
)

# Select the requested columns
vy_columns = [
    "LoggTogID",
    "actual_arrival_time",
    "actual_departure_time",
    "scheduled_arrival_time",
    "scheduled_departure_time",
    "departure_delay",
]

vy = vy[vy_columns]

# Left join: all rows from df_terminal_tios are retained
df_terminal_tios_vy = df_terminal_tios.merge(
    vy,
    on="LoggTogID",
    how="left",
    validate="one_to_one",
)
```

```python
# Display 5 random rows
display(df_terminal_tios_vy.sample(n=5, random_state=37))
```

### Terminallogg $\leftarrow$ TIOS delay report $\leftarrow$ Vy Alnabru timings $\leftarrow$ Årsakskode mapping

*Comment:* The aarsakskode_mapping table works like a dimension table for the Terminallogg. This enables comparisons of the internal delay source reason to the "official" one made by Bane NOR (triangulating of data)

```python
# Create a lookup table from the dimension table
aarsakskode_mapping = (
    df_aarsakskodeliste[
        [
            "Nedtrekks-tekst",
            "Tilhørende YtelsesordningKode",
        ]
    ]
    .copy()
    .rename(columns={
        "Nedtrekks-tekst": "Årsakskode - terminal",
    })
)

# Standardize the join columns
df_terminal_tios_vy["Årsakskode - terminal"] = (
    df_terminal_tios_vy["Årsakskode - terminal"]
    .astype("string")
    .str.strip()
)

aarsakskode_mapping["Årsakskode - terminal"] = (
    aarsakskode_mapping["Årsakskode - terminal"]
    .astype("string")
    .str.strip()
)

# Check that each terminal reason has at most one performance-code mapping
mapping_check = (
    aarsakskode_mapping
    .groupby("Årsakskode - terminal")[
        "Tilhørende YtelsesordningKode"
    ]
    .nunique(dropna=True)
)

conflicting_mappings = mapping_check[mapping_check > 1]

if not conflicting_mappings.empty:
    print("Conflicting mappings found:")
    display(conflicting_mappings)
else:
    print("Each terminal reason has at most one performance-code mapping.")

# Remove completely identical mapping rows, if any
aarsakskode_mapping = aarsakskode_mapping.drop_duplicates()

# Add the corresponding performance code
df_terminal_tios_vy_aarsak = df_terminal_tios_vy.merge(
    aarsakskode_mapping,
    on="Årsakskode - terminal",
    how="left",
    validate="many_to_one",
)
```

```python
# Display 5 random rows
display(df_terminal_tios_vy_aarsak.sample(n=20, random_state=37))
```
