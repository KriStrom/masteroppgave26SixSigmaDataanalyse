# Import and export of data
The following notebook imports the raw data from the unstructured, original datasets (excel files) and converts the data into structured .csv-files for the analysis. No data transformation is done here, but the names of employees are filtered out.

## Package imports

```python
import pandas as pd
```

## Dataset - Terminal data R25

```python
# Import
df_alna_25 = pd.read_excel("../data/raw/Terminalrapportering R25_Alnabru (GAMMEL).xlsm",
                   sheet_name="Terminallogg R25", header=12, usecols="A:U,Z:AA")
df_alna_25.head()
```

```python
# Export to csv
df_alna_25.to_csv("../data/alnabrulogg_R25.csv", index=False, sep=";", encoding="utf-8-sig")
```

## Dataset - Terminal data R26

```python
# Import
df_alna_26 = pd.read_excel("../data/raw/Terminalrapportering R26_Alnabru kopi.xlsm",
                   sheet_name="Terminallogg R26", header=12, usecols="A:U,Z:AA")
df_alna_26.head()
```

```python
# Export to csv
df_alna_26.to_csv("../data/alnabrulogg_R26.csv", index=False, sep=";", encoding="utf-8-sig")
```

## Dataset - TIOS

```python
# Imports and initial column/row selection

df_tios = (pd.read_html("../data/raw/ExcelReport-6f2dcb04-d1bc-4d7e-b997-cbd3a93fe1b3.xls", match="Tognr", header=0, encoding="utf-8")[0]
      .drop(columns=["Unnamed: 3", "Unnamed: 4"]))
df_tios = df_tios[~df_tios["Dato"].str.startswith("Sum", na=False)]    # Remove the summary row at the bottom

df_tios.head()
df_tios.tail()
```

```python
# Export to csv
df_tios.to_csv("../data/tios_Alnabru_delays_R25_R26.csv", index=False, sep=";", encoding="utf-8-sig")
```

## Dataset - TIOS "Operating Train Route (OTR)" from Vy

```python
# Import
df_tios_time = pd.read_csv("../data/raw/bronze_tp_operating_train_routes_ALB.csv", na_values=["null"],
                           usecols=lambda c: c not in ["line_id", "train_type", "train_kind"])   # helt tomme kolonner

df_tios_time.head()
df_tios_time.tail()
```
